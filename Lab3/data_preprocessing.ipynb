```python
# 1. Essential Library Imports
from pathlib import Path
import numpy as np
import pandas as pd
from sklearn.preprocessing import LabelEncoder, MinMaxScaler, OneHotEncoder, StandardScaler

# Configure Pandas display formatting
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)



# Configure dynamic relative paths (works from repo root or Lab3 directory)
DATA_DIR = Path("datasets") if Path("datasets").exists() else Path("../datasets")
OUTPUT_DIR = Path("Lab3/outputs") if Path("Lab3").exists() else Path("outputs")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

print(f"Data Directory  : {DATA_DIR.resolve()}")
print(f"Output Directory: {OUTPUT_DIR.resolve()}")
```

```python
# 2. Dataset Loading with Missing File Verification
dataset_path = DATA_DIR / "Placement_Data_Full_Class.csv"

# Validate file presence before reading
if not dataset_path.exists():
    raise FileNotFoundError(f"Missing required dataset: {dataset_path}. Please ensure datasets/Placement_Data_Full_Class.csv exists.")

# Load CSV dataset into DataFrame
df = pd.read_csv(dataset_path)
print(f"Dataset successfully loaded from: {dataset_path}")
print(f"Total Rows: {df.shape[0]}, Total Columns: {df.shape[1]}")
```

```python
# 3. Structural Inspection: head(), tail(), and shape
print("--- First 5 Records (head) ---")
display(df.head())

print("--- Last 5 Records (tail) ---")
display(df.tail())

print(f"--- Dataset Dimensions (shape) ---")
print(f"Rows: {df.shape[0]}, Columns: {df.shape[1]}")
```

```python
# 4. Data Types and Memory Inspection using info()
print("--- Dataset Information ---")
df.info()

print("--- Data Types Summary ---")
dtypes_summary = pd.DataFrame({
    'Column': df.columns,
    'Data Type': df.dtypes.astype(str),
    'Non-Null Count': df.notnull().sum(),
    'Null Count': df.isnull().sum()
}).reset_index(drop=True)
display(dtypes_summary)
```

```python
# 5. Statistical Summaries using describe()
print("--- Numerical Features Statistical Summary ---")
display(df.describe().round(2).T)

print("--- Categorical Features Statistical Summary ---")
display(df.describe(include=['object']).T)
```

```python
# 6. Duplicate Records Verification
duplicate_count = df.duplicated().sum()
print(f"Duplicate records found: {duplicate_count}")

if duplicate_count > 0:
    df = df.drop_duplicates()
    print(f"Removed duplicates. Updated shape: {df.shape}")
else:
    print("Verification passed: No duplicate rows present in the dataset.")
```

```python
# 7. Missing Value Identification and Percentage Calculation
missing_counts = df.isnull().sum()
missing_percentages = (missing_counts / len(df)) * 100

missing_table = pd.DataFrame({
    'Missing Count': missing_counts,
    'Missing Percentage (%)': missing_percentages.round(2)
})

# Filter columns that contain at least one missing value
missing_columns = missing_table[missing_table['Missing Count'] > 0]
print("--- Columns with Missing Values ---")
display(missing_columns)
```

```python
# 8. Domain-Appropriate Missing Value Handling
# In this dataset, 'salary' has missing values for students whose placement status is 'Not Placed'.
# Let us verify if all missing salary values strictly correspond to 'Not Placed' students:
unplaced_mask = df['status'] == 'Not Placed'
all_unplaced_are_null = df.loc[unplaced_mask, 'salary'].isnull().all()
placed_null_count = df.loc[~unplaced_mask, 'salary'].isnull().sum()

print(f"Do all 'Not Placed' students have null salary? {all_unplaced_are_null}")
print(f"Number of 'Placed' students with missing salary: {placed_null_count}")

# Handling: Since non-placed students received no salary offer, imputation with 0.0 is statistically sound.
df_clean = df.copy()
df_clean['salary'] = df_clean['salary'].fillna(0.0)

print(f"Missing values remaining after domain imputation: {df_clean.isnull().sum().sum()}")
```

```python
# 9. Removing Irrelevant Identifier Columns
# 'sl_no' is an arbitrary sequential row counter with no predictive correlation to placement outcomes.
if 'sl_no' in df_clean.columns:
    df_clean = df_clean.drop(columns=['sl_no'])
    print("Removed column: 'sl_no'")
    print(f"Updated DataFrame shape: {df_clean.shape}")
```

```python
# 10. Categorical Feature Encoding
# Separate categorical columns into binary and multi-class categories
binary_cols = ['gender', 'ssc_b', 'hsc_b', 'workex', 'status']
nominal_cols = ['hsc_s', 'degree_t', 'specialisation']

print("Applying Label Encoding for Binary Features:")
label_encoders = {}
for col in binary_cols:
    if col in df_clean.columns:
        le = LabelEncoder()
        df_clean[col] = le.fit_transform(df_clean[col])
        label_encoders[col] = le
        classes_mapping = {cls: idx for idx, cls in enumerate(le.classes_)}
        print(f"  - '{col}': {classes_mapping}")

print("Applying One-Hot Encoding for Multi-Class Nominal Features:")
print(f"  Nominal columns: {nominal_cols}")
df_encoded = pd.get_dummies(df_clean, columns=nominal_cols, drop_first=True, dtype=int)

print(f"Shape after One-Hot Encoding: {df_encoded.shape}")
display(df_encoded.head())
```

```python
# 11. Demonstration of Feature Scaling (Normalization & Standardization)
numeric_score_cols = ['ssc_p', 'hsc_p', 'degree_p', 'etest_p', 'mba_p']

# Standardization (StandardScaler: mean=0, variance=1)
std_scaler = StandardScaler()
df_standardized = df_encoded.copy()
df_standardized[numeric_score_cols] = std_scaler.fit_transform(df_encoded[numeric_score_cols])

# Normalization (MinMaxScaler: scales values to range [0, 1])
minmax_scaler = MinMaxScaler()
df_normalized = df_encoded.copy()
df_normalized[numeric_score_cols] = minmax_scaler.fit_transform(df_encoded[numeric_score_cols])

print("--- Original Score Columns (First 3 Rows) ---")
display(df_encoded[numeric_score_cols].head(3))

print("--- Standardized Score Columns (Mean ~ 0, Std ~ 1) ---")
display(df_standardized[numeric_score_cols].head(3))

print("--- Normalized Score Columns (Range [0, 1]) ---")
display(df_normalized[numeric_score_cols].head(3))
```

```python
# 12. Before and After Preprocessing Comparison Summary
summary_before = {
    'Total Rows': df.shape[0],
    'Total Columns': df.shape[1],
    'Total Missing Cells': int(df.isnull().sum().sum()),
    'Categorical Columns': len(df.select_dtypes(include=['object']).columns),
    'Numerical Columns': len(df.select_dtypes(include=['int64', 'float64']).columns)
}

summary_after = {
    'Total Rows': df_encoded.shape[0],
    'Total Columns': df_encoded.shape[1],
    'Total Missing Cells': int(df_encoded.isnull().sum().sum()),
    'Categorical Columns': len(df_encoded.select_dtypes(include=['object']).columns),
    'Numerical Columns': len(df_encoded.select_dtypes(include=['int64', 'float64', 'int32', 'int8', 'uint8']).columns)
}

comparison_df = pd.DataFrame([summary_before, summary_after], index=['Raw Dataset', 'Preprocessed Dataset'])
print("--- Before vs After Preprocessing Summary ---")
display(comparison_df)

# Export preprocessed dataset to outputs directory
output_csv_path = OUTPUT_DIR / "preprocessed_placement.csv"
df_encoded.to_csv(output_csv_path, index=False)
print(f"saved preprocessed placement data to: {output_csv_path}")
```
